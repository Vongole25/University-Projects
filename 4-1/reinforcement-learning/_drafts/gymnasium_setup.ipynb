```python
env = gym.make("CarRacing-v3", render_mode="human")  # render_mode="rgb_array"도 가능

# 환경 확인
obs, info = env.reset()

for _ in range(1000):
    action = env.action_space.sample()  # 랜덤 행동
    obs, reward, terminated, truncated, info = env.step(action)
    if terminated or truncated:
        obs, info = env.reset()

env.close()
```

```python
import gymnasium as gym
import pygame
import numpy as np

# 환경 생성
env = gym.make("CarRacing-v3", render_mode="human")
obs, info = env.reset()

pygame.init()
window = pygame.display.set_mode((1000, 1000))  # 더미 창

def get_action():
    keys = pygame.key.get_pressed()
    steer, gas, brake = 0.0, 0.0, 0.0
    if keys[pygame.K_LEFT]: steer = -1.0
    if keys[pygame.K_RIGHT]: steer = 1.0
    if keys[pygame.K_UP]: gas = 1.0
    if keys[pygame.K_DOWN]: brake = 1.0
    return np.array([steer, gas, brake], dtype=np.float32)

try:
    running = True
    while running:
        for event in pygame.event.get():
            if event.type == pygame.QUIT or (event.type == pygame.KEYDOWN and event.key == pygame.K_ESCAPE):
                running = False

        action = get_action()
        obs, reward, terminated, truncated, info = env.step(action)
        if terminated or truncated:
            obs, info = env.reset()

finally:
    env.close()
    pygame.quit()
    print("✅ 창 정상 종료됨")
```
