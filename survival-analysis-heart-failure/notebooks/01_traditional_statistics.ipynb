## Chi-Squared Test, Mann-Whitney 검정, Sharpiro-wilk test

```python
import scipy.stats as stats

group1 = data[data['DEATH_EVENT'] == 1]
group2 = data[data['DEATH_EVENT'] == 0]

def mann_whitney_test(var):
    stat, p = stats.mannwhitneyu(group1[var], group2[var], alternative='two-sided')
    return pd.Series({'Statistic': stat, 'p-value': p})

def chi_square_test(var):
    contingency_table = pd.crosstab(data[var], data['DEATH_EVENT'])
    stat, p, dof, expected = stats.chi2_contingency(contingency_table)
    return pd.Series({'Statistic': stat, 'p-value': p})

continuous_vars = ['age', 'creatinine_phosphokinase', 'ejection_fraction', 'platelets', 
                   'serum_creatinine', 'serum_sodium', 'time']
con_cat_vars = ['age', 'creatinine_phosphokinase', 'ejection_fraction', 'platelets', 
                   'serum_creatinine', 'serum_sodium', 'time', 'anaemia', 'diabetes', 'high_blood_pressure', 'sex', 'smoking']

categorical_vars = ['anaemia', 'diabetes', 'high_blood_pressure', 'sex', 'smoking']

continuous_test_results = pd.DataFrame({var: mann_whitney_test(var) for var in continuous_vars}).T
categorical_test_results = pd.DataFrame({var: chi_square_test(var) for var in con_cat_vars}).T

continuous_test_results, categorical_test_results
```

```python
def shapiro_wilk_test(var):
    stat, p = stats.shapiro(data[var])
    return pd.Series({'Statistic': stat, 'p-value': p})

# Correlation matrix
correlation_matrix = data.corr()

# Apply the Shapiro-Wilk test for continuous variables
shapiro_test_results = pd.DataFrame({var: shapiro_wilk_test(var) for var in con_cat_vars}).T

# Display the results
shapiro_test_results, correlation_matrix
```
